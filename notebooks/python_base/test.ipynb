```python
from fasta import read_fasta

# 普通 .fa
records1 = list(read_fasta("/home/zzc/projects/biopython/src/data/test.fa"))
print("普通:", records1)
assert records1 == [("seq1", "ACGU"), ("seq2", "ACGUU")]

# 压缩 .gz
records2 = list(read_fasta("/home/zzc/projects/biopython/src/data/test.fa.gz"))
print("压缩:", records2)
assert records2 == [("seq1", "ACGU"), ("seq2", "ACGUU")]

# 两种结果一致
assert records1 == records2

print("通过")
```
