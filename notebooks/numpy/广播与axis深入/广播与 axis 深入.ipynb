# 广播的三条规则
维度相等：比如都是 3。
如：（4,3）对（4,3）→（4,3）
其中一个是 1：比如 3 和 1，可以扩展成 3。
如：（4,1）对（1,3）→ （4,3）
其中一方缺失：缺失的那一方会被当成 1，然后按第 2 条扩展。
（4,3）对（  3，）→   （4,3）
A:  2  3  4
B:     3  1
# 小tips
1. numpy中的sum用法：参考mean，mean是求平均值，但是sum就是求那一维的和。
keepdims参数：是否保持维度。（用shape看一看）
2. 改一改sum中的axis和keepdimds，看看归一化操作还能行吗
3. 中心化一般叫对axis=k做中心化。
操作：
1.mean求axis=k的均值，注意keepdims
2.原矩阵-均值，得到的就是已经中心化的矩阵
# 功能大全
## 广播：
让不同形状数组能直接运算。
写代码更短，运行更快。
## 中心化：
去掉均值偏移。
PCA、协方差、线性回归等预处理常用。
## 归一化 / 标准化：
统一特征尺度。
防止大数值特征主导。
加速训练，提高模型效果。
## 
中心化是标准化的一部分。
标准化 = 中心化 + 缩放。
归一化这个词要看上下文，可能是 Min-Max，也可能是标准化。

```python
import numpy as np 
import random
```

```python
counts = np.array([
    [1, 2, 3],
    [4, 5, 6],
])
```

```python
#以下是归一化操作（把某一维的和变成1）
row_sum = counts.sum(axis=1, keepdims=True)
print(row_sum)
```

```python
norm = counts / row_sum
print(norm)
print(norm.shape)
print(norm.sum(axis=1))
#归一化操作结束
```

```python
#每列中心化
col_mean = counts.mean(axis=0)
centered = counts  -col_mean
print(centered)
print(centered.mean(axis=0))
```

```python
import numpy as np

X = np.array([[1, 100],
              [2, 200],
              [3, 300]])

# 中心化
X_centered = X - X.mean(axis=0)

# 标准化
X_standard = (X - X.mean(axis=0)) / X.std(axis=0)

# Min-Max 归一化
X_minmax = (X - X.min(axis=0)) / (X.max(axis=0) - X.min(axis=0))

print("中心化:\n", X_centered)
print("标准化:\n", X_standard)
print("Min-Max 归一化:\n", X_minmax)
```
