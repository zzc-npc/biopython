# random 的使用
## 设置随机数种子
rng = np.random.default_rng(42)
## 根据随机数种子使用分布
旧写法（np.random.xxx）	        新写法（rng.xxx）
np.random.rand(2, 3)	        rng.random((2, 3))            #[0,1)均匀分布
np.random.randn(2, 3)	        rng.standard_normal((2, 3))   #标准正态分布
np.random.randint(1, 7, 10)	    rng.integers(1, 7, 10)        #[1,7)整数
np.random.normal(0, 1, 5)	    rng.normal(0, 1, 5)           #正态分布
np.random.uniform(0, 1, 5)	    rng.uniform(0, 1, 5)          #均匀分布
np.random.choice([1,2,3], 5)	rng.choice([1,2,3], 5)        #随机选择
np.random.shuffle(x)	        rng.shuffle(x)                #原地打乱

# 小tip
1. 矩阵中直接独有的：如矩阵[1,2,3]>1返回的是矩阵[False,Ture,Ture]
2. a[0]和a[0,2],一个用了一维的索引，一个用了二维的；对于[4,3]矩阵，0先找4那一维的，2找3那一维的，从左向右
3. 像normal这样的，最后一个参数决定了生成多少个，5→5个+一维；（4,3）→4x3+二维。不过这里尽量传元组

```python
rng = np.random.default_rng(42)
a =rng.uniform(1, 7,(4,3,2))
print(a[0],"\n",a[:,-1],"\n",a[1:3,0:2])
```

```python
# 用布尔筛选，从 counts 中取出第 1 列大于某阈值的那些行。
mark1 = a[:,:,0] >0
print(mark1.shape)
print(a.shape)
print(a[mark1].shape)
print(np.array_equal(a,a[mark1].reshape(4,3,2)))
```
