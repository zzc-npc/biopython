# tips
1. 基础聚合
a = np.array([50, 120, 80, 200, 30])
a.sum()      # 480
a.mean()     # 96.0
a.std()      # 标准差
a.min()      # 30
a.max()      # 200
a.argmin()   # 4，最小值在索引4
a.argmax()   # 3，最大值在索引3
2. 排序：
scores = np.array([0.5, 0.9, 0.1, 0.7])
np.sort(scores)              # [0.1 0.5 0.7 0.9]，返回新数组
np.argsort(scores)           # [2 0 3 1]，返回排序后的原索引
scores[np.argsort(scores)]   # 等价于 np.sort(scores)
order = np.argsort(scores)[::-1]   # [1 3 0 2]
scores[order]                      # [0.9 0.7 0.5 0.1]
3. top-k：
k = 2
top_idx = np.argsort(scores)[::-1][:k]   # 前2大的索引
top_val = scores[top_idx]
print(top_idx, top_val)                   # [1 3] [0.9 0.7]
4. 配合 axis：
m = np.array([
    [1, 9, 3],
    [4, 2, 8],
])
m.max(axis=1)      # [9 8]，每行最大值
m.argmax(axis=1)   # [1 2]，每行最大值所在的列索引
5.  中位数和百分位：
a = np.array([10, 20, 30, 40, 100])
np.median(a)            # 30
np.percentile(a, 90)    # 90分位数
6. max(axis=1),意味着只在axis=1这一维中取最大值
如：（4,5,2）→（4,2），矩阵中每一个最值都是对应axis=1中的最值
